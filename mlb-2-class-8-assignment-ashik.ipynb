```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = pd.read_csv('/kaggle/input/datasets/lovishbansal123/titanic-dataset/Titanic-Dataset.csv')
iris = pd.read_csv('/kaggle/input/datasets/himanshunakrani/iris-dataset/iris.csv')
tips = pd.read_csv('/kaggle/input/datasets/saurabhbadole/restaurant-tips-dataset/tips.csv')
```

# Visualize at least 1 graph of each category. You can take help from the reference notebook

# Univariate Analysis

# 1. Pie chart

```python
titanic
```

```python
class_count = titanic['Pclass'].value_counts().sort_index()
class_label = ['1st class','2nd class','3rd class']
class_color = ['red','green','blue']
plt.title('Titanic Pclass segmentation')
plt.pie(class_count,
        labels = class_label,
       colors = class_color,
       autopct = "%0.2f%%")
plt.savefig('Pclass_piechart.png')
plt.show()
```

# 2. Count plot

```python
tips
```

```python
sns.countplot(data = tips, x = 'sex')
plt.title('Countplot by Gender on Tips_Dataset')
plt.show()
```

# 3. Kde + Rug plot

```python
sns.kdeplot(titanic['Age'])
sns.rugplot(titanic['Age'])
plt.title('Count on different Age & find density by rugplot')

plt.show()
```

# 4. Violin plot

```python
sns.violinplot(data = titanic, y = 'Pclass')
plt.title('Violin Plot for Pclass')
plt.show()
```

# 5. Box plot

```python
sns.boxplot(data = tips, x = 'tip')
plt.title('Box plot on tip of Client')
plt.show()
```

# 6. Boxen plot

```python
sns.boxenplot(data = titanic, x = 'Fare')
plt.title('Boxen plot for Fare on TitanicnDataset')
plt.show()
```

# 7. Histogram

```python
sns.histplot(titanic['Age'])
plt.title('Age of all Passengers')
plt.show()
```

# Bivariate Analysis

# 1. Scatter plot

```python
plt.scatter(tips['total_bill'],tips['tip'])
plt.xlabel('total_bill')
plt.ylabel('tip')
plt.title('Scatter plot on Total_bill and Tip')
plt.show()
```

```python
sns.scatterplot(data = titanic, x = 'Age', y = 'Fare')
plt.title('Scatter plot on age and fare')
plt.show()
```

# 2. Line plot

```python
sns.lineplot(data=tips, x="total_bill", y="tip")

plt.title("Total Bill vs Tip")
plt.show()
```

# 3. Joint plot

```python
sns.jointplot(data = tips, x = 'total_bill', y = 'tip')
plt.title('Joint plot between total_bill and tip')
plt.show()
```

# 4. Hexbin plot

```python
plt.hexbin(iris['petal_length'], 
           iris['petal_width'], 
           gridsize=25, cmap='Blues')

plt.colorbar(label='count in bin')
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("Hexbin Plot: Petal Length vs Petal Width")

plt.show()
```

# 5. 2D kde plot

```python
sns.kdeplot(data = iris, 
            x='petal_length', 
            y='petal_width', 
            fill = True)
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("2D KDE Plot between Petal Length and Petal Width")
plt.show()
```

# 6. Box plot

```python
sns.boxplot(data = titanic, 
            y = 'Age', 
            x = 'Survived')
plt.title('Bivariate boxplot for Age & Survived')
plt.show()
```

# 7. Violin plot

```python
sns.violinplot(data = titanic, 
               y = 'Pclass', 
               x ='Survived')
plt.title('Violin Plot for Pclass')
plt.show()
```

# 8. Swarm plot

```python
sns.swarmplot(data = tips, 
              x='total_bill',
              y='day',
              hue='day')
plt.title('Swarm plot between total bill and day of Tips dataset')
plt.show()
```

# 9. Count plot

```python
sns.countplot(data = titanic, 
              x = 'Pclass', 
              hue = 'Survived')
plt.title('Count plot on pclass and survived passenger')
plt.show()
```

# Multivariate Analysis

# 1. Scatter plot

```python
sns.scatterplot(data = iris, 
                x = 'petal_length',
                y = 'petal_width',
                hue = 'species')
plt.title('Scatter plot on petal length and petal width to find species')
plt.show()
```

# 2. Pair plot

```python
sns.pairplot(titanic, hue = 'Survived')
plt.show()
```

# 3. Heatmap

```python
sns.heatmap(iris.select_dtypes(include = 'number').corr(), cmap = 'coolwarm', annot = True)
plt.show()
```

# 4. FacetGrid plot

```python
g = sns.FacetGrid(tips,col = 'day',row = 'time')
g.map_dataframe(sns.histplot,x = 'total_bill')
plt.show()
```

# 5. PairGrid

```python
g = sns.PairGrid(iris,hue='species')
g.map_diag(sns.histplot)
g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot)
g.add_legend()
g.map_offdiag(sns.scatterplot)
plt.show()
```

# 6. Joint plot

```python
sns.jointplot(data = titanic, 
              x = 'Age', 
              y = 'Fare', 
              hue = 'Survived')
plt.show()
```

# 7. Boxen plot

```python
sns.boxenplot(data= titanic, 
              x='Pclass', 
              y='Age', 
              hue='Sex')

plt.title('Boxen plot on Age Distribution by Passenger Class and Sex')
plt.show()
```
